```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# STEP 1: DATA CREATION
data = {
    'Gender': ['Male', 'Female', 'Male', 'Male', 'Male'],
    'Married': ['Yes', 'No', 'Yes', np.nan, 'Yes'],
    'Dependents': ['0', '1', '3+', '2', '0'],
    'Education': ['Graduate', 'Not Graduate', 'Graduate', 'Graduate', 'Graduate'],
    'Self_Employed': ['No', 'Yes', 'No', 'No', 'No'],
    'ApplicantIncome': [5000, 3000, 4000, 6000, 5000],
    'CoapplicantIncome': [1500, 0, 2000, 1000, 1500],
    'LoanAmount': [150, 100, np.nan, 200, 150],
    'Loan_Amount_Term': [360, 360, 360, np.nan, 360],
    'Credit_History': [1, 1, np.nan, 1, 1],
    'Property_Area': ['Urban', 'Rural', 'Semiurban', 'Urban ', 'Urban']
}

df = pd.DataFrame(data)
print("Dataset created successfully!\n")
display(df)

# STEP 2: INITIAL DATA OVERVIEW
print("\n" + "=" * 60)
print("INITIAL DATA OVERVIEW")
print("=" * 60)
df.info()

# STEP 3: CATEGORICAL DATA CLEANING (Strip whitespaces first)
categorical_columns = ['Gender', 'Married', 'Education', 'Self_Employed', 'Property_Area']
for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip().str.capitalize()
    # Restore actual NaNs that were converted to 'Nan' string
    df[col] = df[col].replace('Nan', np.nan)

# STEP 4: HANDLING MISSING VALUES
print("\n" + "=" * 60)
print("MISSING VALUES BEFORE CLEANING")
print("=" * 60)
print(df.isnull().sum())

# Fill categorical columns using MODE
for col in ['Gender', 'Married', 'Self_Employed']:
    df[col] = df[col].fillna(df[col].mode()[0])

# Fill LoanAmount using MEDIAN
df['LoanAmount'] = df['LoanAmount'].fillna(df['LoanAmount'].median())

# Fill Loan_Amount_Term using MODE
df['Loan_Amount_Term'] = df['Loan_Amount_Term'].fillna(df['Loan_Amount_Term'].mode()[0])

# Fill Credit_History using MODE
df['Credit_History'] = df['Credit_History'].fillna(df['Credit_History'].mode()[0])

# STEP 5: DATA TYPE CONVERSION & DEPENDENTS CLEANING
# Safely map '3+' to 3, fill missing, and convert to integer
df['Dependents'] = df['Dependents'].replace('3+', 3).fillna(df['Dependents'].mode()[0]).astype(int)

print("\n" + "=" * 60)
print("MISSING VALUES AFTER CLEANING")
print("=" * 60)
print(df.isnull().sum())

# STEP 6: DUPLICATE REMOVAL
initial_rows = df.shape[0]
df.drop_duplicates(inplace=True)
removed_rows = initial_rows - df.shape[0]
print(f"\nDuplicate rows removed: {removed_rows} (Current rows: {df.shape[0]})")

# STEP 7: SCALING & STANDARDIZATION
min_max_scaler = MinMaxScaler()
scale_columns = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount']
df[scale_columns] = min_max_scaler.fit_transform(df[scale_columns])

standard_scaler = StandardScaler()
df[['Credit_History']] = standard_scaler.fit_transform(df[['Credit_History']])

# STEP 8: FINAL RESULTS
print("\n" + "=" * 60)
print("FINAL CLEANED DATA")
print("=" * 60)
display(df)

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)
display(df.describe())
```
